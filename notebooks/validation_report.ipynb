# Nine subjects walking vs Visual3D: checking that P1's agreement wasn't a fluke

This notebook reports on the following key findings about our reanalysis:

1. **One set of parameters, every trial, both legs.** 251 trials from 9 subjects (0.65–1.94 m/s treadmill walking) run with identical settings: 500 trial × leg comparisons, 1500 joints.
2. **How big the error is.** Median torque RMS vs Visual3D is 2.48 N m at the ankle, 3.75 at the knee and 11.30 at the hip. 82% of the 1000 ankle + knee comparisons are under 5 N m, and 94% of the 500 hip comparisons are under 20 N m.
3. **Flat with speed.** As a fraction of each trial's own Visual3D peak torque, the median error is 1.9% ankle, 4.6% knee, 10.6% hip, and it doesn't grow at the fast end.
4. **The hip is worst, and I think that's the hip centre, not the physics.** The hip is the worst joint for every subject, and its median is roughly 5× the ankle's. The one outlier is p2's left ankle (7.78 N m vs 2.30 on the right), and its excess is in the frontal plane.
5. **Nothing failed.** 0 of 251 trials raised; 1 (p6 trial 25, belt 2 peaks at 5 N) is excluded before any dynamics run, which is 2 of 502 comparisons.
